```python
#built in modules

import math
print(math.sqrt(49))
print(math.sin(math.pi))
print(f"ceil is : {math.ceil(2.3)}")
```

```python
import random
print(random.randint(1,78))

print(f"random float is :{random.random()}")
print(f"random choice  is :{random.choice([2, 4,5 ,0, 5])}")
print(f"random shuffle  is :{random.shuffle(['a', 'b' , 'c', 'd'])}")
```
