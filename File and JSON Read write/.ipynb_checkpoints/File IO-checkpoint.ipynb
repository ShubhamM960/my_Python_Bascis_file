```python
# mode = 'r --> read only
# mode = 'w' --> overwrite only( and create new if not exist)
# mode = 'a' --> only append file
# mode = 'r+' --> read and write
# mode = 'w+' --> read and overwrite (overwrite existing file or create new if not exist)
# mode = 'a+' --> read and appeand(create new if not exist )
```

```python
# mode = 'r' --> read only
myFile = open('test.txt', 'r')
myFile.seek(0)
myFile.read()
myFile.seek(0)
myFile.readlines()
```

```python
# mode = 'r' --> read only
with open('test.txt', mode = 'r') as myFile:
    myFile.seek(0)
    print(myFile.read())
```

```python
# mode = 'a' --> only append file
myFile = open('test.txt', 'a')
myFile.write('\nappend firstline')

myFile = myFile = open('test.txt', 'r')
myFile.read()
```

```python
# mode = 'a' --> read only
with open('test.txt', mode = 'a') as myFile:
    myFile.write('\nwrite my line using with open')


with open('test.txt', mode = 'r') as myFile:
    myFile.seek(0)
    print(myFile.read())
```

```python
# mode = 'w' --> overwrite file (create if not exist)
myFile = open('test.txt', 'w')
myFile.write('overwrite firstline')

myFile = myFile = open('test.txt', 'r')
myFile.seek(0)
yFile.read()
```

```python
# mode = 'w' --> read only
with open('test.txt', mode = 'w') as myFile:
    myFile.write('overwrite my line using with open')


with open('test.txt', mode = 'r') as myFile:
    myFile.seek(0)
    print(myFile.read())
```

```python
# mode = 'a+' --> read and appeand(create new if not exist )
myFile = open('test.txt', 'a+')
myFile.write('\nappend firstline')
myFile.seek(0)
myFile.read()
```

```python
# mode = 'w+' --> read and overwrite (overwrite existing file or create new if not exist)
myFile = open('test.txt', 'w+')
myFile.write('overwrite firstline')
myFile.seek(0)
myFile.read()
```

```python
#When you have finished using a file, it is always good practice to close it.
myFile.close()
```
