```python
# lets create a file to json data 'myJsonFile.json' and store initial data

import json

jsondata = { 'name' : 'ssm', 'profession' : 'SWE', 'Hobby' : 'Cricket'}

with open('myJsonFile.json', mode = 'w') as json_file :
    json.dump( jsondata,json_file, indent=4 )
```

```python
#read the content of the data
with open('myJsonFile.json', mode = 'r') as json_file :
    #print(json.load( json_file ))
    data = json.load( json_file )
```

```python
#update some content
newdata = {'location' : 'HYD', 'mobile' : '9945723'}

data.update(newdata)

with open('myJsonFile.json', mode = 'w+') as json_file :
    json.dump( data,json_file, indent=4 )
```
