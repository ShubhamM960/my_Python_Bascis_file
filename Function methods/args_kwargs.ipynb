```python
# *args  --> Tuple
def args_fun(*args):
    for x in args :
        print(x)


args_fun(10,20, 49, 89)
```

```python
# **kwargs --> Dictionary
def kargs_fun(**kwargs):
    for x in kwargs :
        print(kwargs[x])


kargs_fun(a=10,b=20, c=49, d=89)
```

```python
def mixed_fun(*args, **kwargs):
    print(f"the values are {args[2]} , from the map is {kwargs['a']} ");

mixed_fun(3,4,2,6,8, a = 12, b= 38, c= 89);
    
```
