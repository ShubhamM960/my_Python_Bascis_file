```python
def evaluate_even_num(num_list):
    
    even_numbers = []
    
    # Go through each number
    for number in num_list:
        # Once we get a "hit" on an even number, we append the even number
        if number % 2 == 0:
            even_numbers.append(number)
        # Don't do anything if its not even
        else:
            pass
    # Notice the indentation! This ensures we run through the entire for loop    
    return even_numbers


#call    
final_lst = evaluate_even_num([4, 12, 3, 5,14,63])
print(final_lst)
```

```python
# *args  --> Tuple
def args_fun(*args):
    for x in args :
        print(x)


args_fun(10,20, 49, 89)
```

```python
# **kwargs --> Dictionary
def kargs_fun(**kwargs):
    for x in kwargs :
        print(kwargs[x])


kargs_fun(a=10,b=20, c=49, d=89)
```
