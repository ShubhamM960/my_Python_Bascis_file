```python
#highr order function : We can pass the function name as a parameter to another method and return the funtion result
```

```python
def add(a, b):
    return a+b
def substarct(a, b):
    return a-b
def multiply(a,b):
    return a*b
def divide(a,b):
    return a/b

def calculate(a, b, funName):
    return funName(a,b)
  
```

```python
calculate(4,5, multiply)
```

```python
calculate(4,5, add)
```

```python
def multiplier(factor):
    def multiply(num):
        return factor*num
    return multiply

fun = multiplier(5) # this will return the inner function isntance
result = fun(2) #so instead of calling using the actual name we just pass the function instance variable to call the method
print(result)
```
