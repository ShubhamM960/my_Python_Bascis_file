```python
there four scopes in python
local
enclosing functions
global
built-in
```

```python
LEGB rule : 
L: Local — Names assigned in any way within a function (def or lambda).

E: Enclosing function locals — Names in the local scope of any and all enclosing functions (def or lambda), from inner to outer.

G: Global (module) — Names assigned at the top-level of a module file, or declared global in a def within the file.

B: Built-in (Python) — Names preassigned in the built-in names module : open, range, SyntaxError,..
```

```python
#L: Local — Names assigned in any way within a function (def or lambda).

def func():
    x=2
    print(x)

lambda x : x**2 

so here 'x' is local

========================

x = 50

def func(x):
    print('x is', x)
    x = 2
    print('now x is now', x)

func(4)# x is', 4
print('x is still', x) #x is still, 50
```

```python
#Enclosing Function
#Enclosing function locals — for an inner function, it check for the variable value in the Local  scope, if found then use it if not 
                           #then it takes the value of global scope outside of the function
name = 'This is a global name'

def greet():
    # Enclosing function
    name = 'Sammy'
    
    def hello():
        print('Hello '+name)
    
    hello()

greet() #O/P : Hello Sammy

=======================================

name = 'This is a global name'

def greet():
    # Enclosing function
    name = 'Sammy'
    
    def hello():
        name = 'takes from hello'
        print('Hello '+name)
    
    hello()

greet() #O/P : takes from hello
```

```python
x = 50

def func():
    global x
    print(f"Because of global x is: {x} ")
    x = 2
    print(f'Ran func(), changed global x to : {x}')


func()
print(f'Value of x (outside of func()) is: {x}')
```
