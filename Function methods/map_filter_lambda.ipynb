```python
#map() function in Python is a built-in to apply a specific function to every item in an Collection (like a list, tuple, or dictionary) 
 #without writing an explicit for loop.

# map(fun_name, Iterable) --> iterable = List/ tuple / dict

#wrap the output with in a list() or tuple() or set
#e.g. : list( map(fun_name, Iterable) )
```

```python
def splicer(str):
    if len(str) % 2 == 0:
        return 'EVEN'
    else:
        return str[0]

str_lst = ['wedf', 'few', '3rtrt', 'r24r3']

list( map(splicer, str_lst) ) # o/p : ['EVEN', 'f', '3', 'r']
##===================================
def squre_fun(num):
    return num**2

num_lst = [3,8,23,12,45]

list( map(squre_fun, num_lst) ) # o/p : [9, 64, 529, 144, 2025]
#================================
string_numbers = ["1", "2", "3", "4"]

# Apply the int() function to every item
result = map(int, string_numbers)

print(list(result)) # o/p : [1,2,3,4] 
```

```python
# filter() : extract elements only those elements from an iterable (like a list, tuple, or set) that satisfy a specific condition
#filter(fun_name, iterable)
#list(filter(fun_name, iterable)) or tuple(filter(fun_name, iterable))
```

```python
def is_even(num):
    return num % 2 == 0

numbers = [1, 2, 3, 4, 5, 6]

# Filter the list
list( filter(is_even, numbers) )
```

```python
#lambda is alternative of a function , contains the expression and return the result without any return statement
#lambda arguments: expression

# arguments = parameters to the function , expression = logic and return statement of the function
```

```python
num_lst = [3,8,23,12,45]
list( map(lambda x : x**2,num_lst ) )
```

```python
list1 = [1, 2, 3]
list2 = [10, 20, 30]

# Add elements from both lists together
summed = map(lambda x, y: x + y, list1, list2)

print(list(summed)) 
```

```python
num_lst = [3, 31, 9, 34, 45]
filtered_list = filter(lambda x : x%2 == 0, num_lst)
print(list(filtered_list))
```
