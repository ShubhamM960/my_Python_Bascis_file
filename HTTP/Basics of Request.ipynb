```python
import requests, bs4

res = requests.get('http://www.example.com')

#print(res.content) #print(res.text) #print(res.json()) 
soup = bs4.BeautifulSoup(res.text,"lxml" ) #'html.parser'
print(soup)
```

```python
# then in the 'soup' variable we can use '.select()' to get the content of specific tag
# it default return a value in "list"

Syntax to pass to the .select() method
=================================================
soup.select('tagName')       ==>         All elements with that tag

soup.select('#some_id')      ==>         The HTML element containing that id 

soup.select('.className')    ==>         All the HTML elements with the CSS class 

soup.select('div span')      ==>         Any elements named <span> that are within an element named <div>
```
