```python
class Book:
    def __init__(self, title, author, pages):
        print("A book is created")
        self.title = title
        self.author = author
        self.pages = pages
    
    #it will call when use print()
    def __str__(self):
        return "Title: %s, author: %s, pages: %s" %(self.title, self.author, self.pages)
    #it will call when use len()
    def __len__(self):
        return int(self.pages)
    #it will call when use del()
    def __del__(self):
        print("A book is destroyed")
```

```python
b = Book('Power of Sub Consious Mind', 'John Murphy', 300)
```

```python
print(b)
```

```python
len(b)
```

```python
del b
```
