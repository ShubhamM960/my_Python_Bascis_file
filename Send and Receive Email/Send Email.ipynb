```python
import smtplib, getpass

smtp_obj = smtplib.SMTP('smtp.gmail.com')
#smtp_obj.ehlo()
smtp_obj.starttls()


user_name = getpass.getpass('Enter Email: ') #user_name = 'ssm73123@gmail.com'
password = getpass.getpass('Enter password: ')
#print(password)
to_addr = 'abc@gmail.com'
subject = input('Enter the Subject : ')
body = input('Enter the message body : ')
msg = 'Subject :'+ subject +'\n' + body

try :
    smtp_obj.login(user_name, password) #print(smtp_obj.login(user_name, password))
    smtp_obj.sendmail(user_name, to_addr, msg)
except :
    print('Please verify your username or password')
else : 
    smtp_obj.quit()
```
